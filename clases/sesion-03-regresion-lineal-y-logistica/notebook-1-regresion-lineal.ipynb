# Clase 5 - Regresión Lineal

Vamos a ajustar una recta a mano (con scikit-learn) sobre datos
simulados de casas, y después probar con varias variables a la vez.

## 1. Regresión lineal simple: ajustar una recta

Simulamos precios de casas a partir de su superficie (m²), con algo de
ruido para que no quede una recta perfecta.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=0)

m2 = rng.uniform(30, 150, 60)
precio = 15000 * m2 + 500000 + rng.normal(0, 300000, 60)

plt.scatter(m2, precio)
plt.xlabel("Superficie (m²)")
plt.ylabel("Precio (pesos)")
plt.title("Precio de casas según superficie")
plt.show()
```

Se nota una tendencia lineal, pero con ruido: dos casas del mismo tamaño no
tienen exactamente el mismo precio. Vamos a buscar la recta que mejor
resume esa tendencia.

```python
from sklearn.linear_model import LinearRegression

X = m2.reshape(-1, 1)   # scikit-learn siempre espera una matriz 2D de features
y = precio

modelo = LinearRegression()
modelo.fit(X, y)

print("w (pendiente):", modelo.coef_[0])
print("b (ordenada al origen):", modelo.intercept_)
```

```python
plt.scatter(m2, precio, label="datos")

m2_ordenado = np.linspace(m2.min(), m2.max(), 100).reshape(-1, 1)
plt.plot(m2_ordenado, modelo.predict(m2_ordenado), color="red", label="recta ajustada")

plt.xlabel("Superficie (m²)")
plt.ylabel("Precio (pesos)")
plt.legend()
plt.show()
```

## 2. Evaluando el modelo: train/test y $R^2$

Para evaluar de verdad necesitamos datos que el modelo no haya visto al entrenar.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

modelo = LinearRegression()
modelo.fit(X_train, y_train)

r2 = modelo.score(X_test, y_test)
print("R^2 en test:", round(r2, 3))
```

```python
predicciones = modelo.predict(X_test)

comparacion = pd.DataFrame({
    "m2": X_test.flatten(),
    "precio_real": y_test,
    "precio_predicho": predicciones.round(0),
})
comparacion.sort_values("m2").head(10)
```

## 3. Regresión lineal múltiple

Una sola variable rara vez alcanza. Agregamos `ambientes` y `antiguedad` al
dataset simulado.

```python
n = 200
m2 = rng.uniform(30, 150, n)
ambientes = rng.integers(1, 6, n)
antiguedad = rng.uniform(0, 50, n)

precio = (
    12000 * m2
    + 800000 * ambientes
    - 15000 * antiguedad
    + 400000
    + rng.normal(0, 250000, n)
)

casas = pd.DataFrame({
    "m2": m2,
    "ambientes": ambientes,
    "antiguedad": antiguedad,
    "precio": precio,
})
casas.head()
```

```python
X = casas[["m2", "ambientes", "antiguedad"]]
y = casas["precio"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

modelo = LinearRegression()
modelo.fit(X_train, y_train)

for feature, coef in zip(X.columns, modelo.coef_):
    print(f"{feature}: {coef:.1f}")
print("b:", round(modelo.intercept_, 1))
print("R^2 en test:", round(modelo.score(X_test, y_test), 3))
```

Los coeficientes tienen sentido con cómo armamos los datos: más m² y más
ambientes suben el precio, más antigüedad lo baja. En un dataset real
nunca vamos a saber los coeficientes "verdaderos" de antemano, por eso
simularlos acá ayuda a entender qué está haciendo el modelo por dentro.
